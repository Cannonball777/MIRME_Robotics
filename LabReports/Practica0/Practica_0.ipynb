### Definición de Nodo y Tópicos en ROS 2.
Un "nodo" es un proceso que realiza cálculos. Los nodos se combinan en un gráfico y se comunican entre sí mediante la transmisión de topics, RPC services y el Parameter Server. Estos nodos están destinados a operar a una escala detallada; un sistema de control de robot generalmente comprenderá muchos nodos. Por ejemplo, un nodo controla un telémetro láser, un nodo controla los motores de la rueda del robot, un nodo realiza la localización, un nodo realiza la planificación de la ruta, un nodo proporciona una vista gráfica del sistema, etc. 

En cambio un tópico puede ser definido como un canal de comunicación donde se envían mensajes de un nodo publicador y a su vez uno o más nodos suscriptores pueden leer el contenido y utilizarlo para convertirlo en una nueva información.

### Tres tipos de interfaces que maneja ROS 2

-  __MENSAJES__ : La comunicación es asíncrona y se utilizan tópicos. Sirven para obtener información en tiempo real.
-  __SERVICIOS__ : Permiten una comunicación síncrona. Se componen de una petición (request) y una respuesta (response). Se utilizan para pedir una acción específica esperando un resultado.
-  __ACCIONES__ : Para tareas que toman tiempo en ejecutarse y requieren retroalimentación. Se utilizan para navegación o manipulación que ejecute el robot de manera prolongada.

### Mi primer nodo

    #!/usr/bin/env python3

    import rclpy
    from rclpy.node import Node


    class  MyNode(Node):
        def __init__(self):    
            super().__init__("mi_nodo")
            self.contador_ = 0
            self.get_logger().info("Bivennidos a ROS 2")
            self.create_timer(1, self.callback_timer)
    
        def callback_timer(self):
            self.contador_ += 1
            self.get_logger().info("Hola " + str(self.contador_))


    def main(args=None):
        rclpy.init(args=args)
        node = MyNode()
        rclpy.spin(node)
        rclpy.shutdown()

    if __name__=="__main__":
        main()

### Nodo publicador de una función senoidal cuyos valores tienen unidades de $\frac{rad}{s}$

El nodo publicador utilza de la biblioteca de math la función seno y el número pi para que cuando se ejecute suba a un tópico llamado __publish_topic_velrad__ una función senoidal cuyos valores están en unidades de $\frac{rad}{s}$. Se creó una sentencia get_logger().info para indicar cuando el nodo se activa. Y también se creó un temporizador que ejecuta la función __sin_publish__ cada segundo. A una variable llamaada __msg__ se le asigna el valor de una función seno con frecuencia de 1 Hz y amplitud 1, el valor del tiempo __t__ va cambiando 0.1 [s] cada 1 segundo.

    #!/usr/bin/env python3
    
    import rclpy
    from rclpy.node import Node
    from math import sin,pi
    
    from std_msgs.msg import Float64
     
    
    class MyPublish(Node):
      def __init__(self):
        super().__init__("publish_node")
        self.t_ = 0.0
        self.freq_ = 1
        self.publisher_ = self.create_publisher(Float64,"publish_topic_velrad", 10)
        self.get_logger().info("Nodo publicador activo")
        self.create_timer(1,self.sin_publish)
    
      def sin_publish(self):
        msg = Float64()
        msg.data = sin(2*pi*self.freq_*self.t_)
        self.publisher_.publish(msg)
        self.get_logger().info("La velocidad en rad/s es: " + str(msg.data))
        self.t_+=0.1
        
    
    def main(args=None):
      rclpy.init(args=args)
      node = MyPublish()
      rclpy.spin(node)
      rclpy.shutdown()
    
    if __name__ == '__main__':
      main()

### Nodo suscriptor y publicador de una función senoidal cuyos valores tienen unidades de RPM

Lo que este nodo hace es ingresar al tópico llamado __publish_topic_velrad__ y toma el valor en $\frac{rad}{s}$ de velocidad. Luego convierte este valor en RPM con el factor de conversión de 9.55 y lo publica en un nuevo tópico llamado __publish_topic_rpm__.

NOTA: se puede observar que en todos los programas hay dos renglones de if al final, sirven para garantizar que la función main() se ejecutará solo cuando se abre directamente y no mediante otro script.

    #!usr/bin/env python3
    
    import rclpy
    from rclpy.node import Node
    
    from std_msgs.msg import Float64
    
    
    
    class SubNode(Node):
      def __init__(self ):
        super().__init__("node_sub")
        self.mensaje_ = 0.0
        self.subcriber_ = self.create_subscription(Float64,"publish_topic_velrad",self.sin_publish_rpm,10)
        self.vel_rpm_motor_ = self.create_publisher(Float64,"publish_topic_rpm",10)
        self.get_logger().info("Nodo subcriptor activo")
    
      def sin_publish_rpm(self,msg):
        self.mensaje_ = msg.data
        # self.get_logger().info("El mensaje es " + str(self.mensaje_))
        self.new_msg_ = Float64()
        self.new_msg_.data = self.mensaje_*9.55
        self.vel_rpm_motor_.publish(self.new_msg_)
        self.get_logger().info("La velocidad en rpm es: "+str(self.new_msg_.data))
    
    def main(args=None):
      rclpy.init(args=args)
      node = SubNode()
      rclpy.spin(node)
      rclpy.shutdown()
    
    if __name__ == "__main__":
      main()
